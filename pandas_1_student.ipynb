## 4. Selecting by Label (`.loc`) and by Position (`.iloc`)

- `df.loc[rows, columns]` uses **labels** (index values and column names). Slices **include** the end.
- `df.iloc[rows, columns]` uses **positions** (0, 1, 2, …). Slices **exclude** the end, like normal Python.

🖥️ **Demo:** rows labelled 0 to 4, three named columns.

```python
df.loc[0:4, ['Name', 'Survived', 'Sex']]
```

🤔 **Predict:** how many rows does `df.loc[0:4]` return? How many does `df.iloc[0:4]` return?

```python
df.iloc[0:4]
```

## 5. Missing Values

🖥️ **Demo:** count missing values in every column.

```python
df.isna().sum()
```

🖥️ **Demo:** show the rows where `Age` is missing.

```python
df[df['Age'].isna()]
```

✏️ **Your turn:** create `df_age` containing only the rows where `Age` is **not** missing. Check how many rows it has. (Hint: there's a method called `notna()`.)

🖥️ **Demo:** instead of dropping, fill missing ages with the median age. We work on a **copy** so `df` stays untouched.

```python
# make a copy of df
df_filled = df.copy()

median_age = df_filled['Age'].median()
df_filled['Age'] = df_filled['Age'].fillna(median_age)
df_filled['Age'].isna().sum()
```

## Recap
| Task | Code |
|---|---|
| Load a CSV | `pd.read_csv("file.csv")` |
| Overview | `.head()`, `.shape`, `.info()`, `.describe()`, `.value_counts()` |
| One / several columns | `df['col']` / `df[['a', 'b']]` |
| Filter rows | `df[df['col'] > x]`, combine with `&`, `\|` and parentheses |
| By label / position | `df.loc[...]` (end included) / `df.iloc[...]` (end excluded) |
| Missing values | `.isna()`, `.notna()`, `.dropna()`, `.fillna()` |
| Rate of a 0/1 column | `df['col'].mean()` |
